```python
from dominate.tags import body, html, p, style

# 创建一个 HTML 文档
doc = html()
with doc.add(body()):
    # 添加样式
    with style(type="text/css"):
        print("""
        [data-after]::after {
          content: attr(data-after);
        }
        """)
    # 创建一个段落元素
    paragraph = p("这是一个段落。")
    # 添加内联样式
    paragraph['style'] = "color: red;"
    # 添加 ::after 伪元素的样式
    paragraph['data-after'] = "content: ' ::after 内容'; color: blue;"

# 打印生成的 HTML
print(doc)
```
